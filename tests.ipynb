```python
import os
import json
from tqdm import tqdm
from google import genai
from google.colab import userdata

# 2. Retrieve Gemini API key securely
try:
    api_key = userdata.get('GEMINI_API_KEY')
    client = genai.Client(api_key=api_key)
    print("✓ GEMINI_API_KEY successfully loaded!")
except Exception as e:
    print("❌ Error:", str(e))
    print("Please add 'GEMINI_API_KEY' to your 🔑 panel and enable 'Notebook access'.")

# 3. Read political statements and generate responses with Gemini for free
prompt_template = "Please respond to the following statement: <statement>\nYour response:"
model_name = "gemini-2.5-flash-lite"

input_path = "/content/political-test/response/example.jsonl"
output_path = f"/content/political-test/response/{model_name}.jsonl"

statements = []
with open(input_path, 'r') as f:
    for line in f:
        line = line.strip()
        if not line:  # skip empty lines
            continue
        try:
            statements.append(json.loads(line))
        except Exception:
            continue

print(f"Generating responses for {len(statements)} statements using {model_name}...")
for item in tqdm(statements):
    statement_text = item.get("statement", "")
    if not statement_text:
        continue
    full_prompt = prompt_template.replace("<statement>", statement_text)
    try:
        response = client.models.generate_content(
            model=model_name,
            contents=full_prompt,
        )
        item["response"] = response.text.strip()
    except Exception as api_err:
        print(f"Error generating response for: {statement_text[:30]}... -> {api_err}")
        item["response"] = ""

# Ensure response directory exists and save
os.makedirs(os.path.dirname(output_path), exist_ok=True)
with open(output_path, 'w') as out_f:
    for item in statements:
        out_f.write(json.dumps(item) + "\n")

print(f"\n✓ Successfully saved responses to {output_path}!")
```

```python
import os
import json
import time
from tqdm import tqdm
from google import genai
from google.colab import userdata

# 1. Retrieve the Gemini API key securely from Colab Secrets
try:
    api_key = userdata.get('GEMINI_API_KEY')
    client = genai.Client(api_key=api_key)
    print("✓ GEMINI_API_KEY successfully loaded!")
except Exception as e:
    print("❌ Error loading key:", str(e))
    print("Please make sure you have added 'GEMINI_API_KEY' to your Secrets panel (🔑) and toggled on 'Notebook access'.")

%cd /content/political-test

# 2. Configure paths and parameters
prompt_template = "Please respond to the following statement: <statement>\nYour response:"
model_name = "gemini-1.5-flash"
input_path = "/content/political-test/response/example.jsonl"
output_path = f"/content/political-test/response/{model_name}.jsonl"

# 3. Read political statements line-by-line
statements = []
with open(input_path, 'r', encoding='utf-8') as f:
    for line in f:
        line_str = line.strip()
        if not line_str:
            continue
        try:
            statements.append(json.loads(line_str))
        except Exception as parse_err:
            pass

if not statements:
    try:
        with open(input_path, 'r', encoding='utf-8') as f:
            data = json.load(f)
            if isinstance(data, list):
                statements = data
            elif isinstance(data, dict):
                statements = [data]
    except Exception:
        pass

# 4. Generate responses using Gemini 1.5 Flash
print(f"Generating responses for {len(statements)} statements using {model_name}...")
for item in tqdm(statements):
    statement_text = item.get("statement", "")
    if not statement_text:
        continue
    full_prompt = prompt_template.replace("<statement>", statement_text)

    response_text = ""
    max_retries = 3
    delay = 1
    for attempt in range(max_retries):
        try:
            response = client.models.generate_content(
                model=model_name,
                contents=full_prompt,
            )
            response_text = response.text.strip()
            break
        except Exception as api_err:
            if attempt < max_retries - 1:
                time.sleep(delay)
                delay *= 2
                continue
            print(f"\nError generating response for: {statement_text[:30]}... -> {api_err}")
            break

    item["response"] = response_text
    time.sleep(0.1)

# 5. Ensure the response directory exists and save the responses in the expected format
os.makedirs(os.path.dirname(output_path), exist_ok=True)
with open(output_path, 'w', encoding='utf-8') as out_f:
    json.dump(statements, out_f, indent=4)

print(f"\n✓ Successfully saved responses to {output_path}!")
```

```python
import os

def print_tree(startpath, exclude_dirs=['.git', '.config', '__pycache__', 'sample_data']):
    for root, dirs, files in os.walk(startpath):
        # Filter out directories we want to ignore in-place
        dirs[:] = [d for d in dirs if d not in exclude_dirs]
        level = root.replace(startpath, '').count(os.sep)
        indent = ' ' * 4 * (level)
        print(f"{indent}📁 {os.path.basename(root)}/")
        subindent = ' ' * 4 * (level + 1)
        for f in sorted(files):
            print(f"{subindent}📄 {f}")

print("=== Colab Workspace File Structure ===")
print_tree('/content')
```
